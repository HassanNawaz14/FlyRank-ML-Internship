## **3. Quick look at the data (2-3 real numbers)**

### **Key Observations (Post-Filtering)**
*   Approximately *3000* unique content pages.
*   About *54.21%* of these pages show a 'down' trend.
*   These declining pages account for around *51.27%* of total impressions.

### **Motivation**
*   Content decline is a **significant phenomenon**.
*   Affects a *notable* portion of content footprint and overall traffic.
*   Justifies deeper analysis into associated signals.


```python
import pandas as pd
import os

# Assert path to data
assert os.path.exists("data/raw/content_refresh_anonymized.csv")
```

```python
# Load the starter CSV
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
df.head(2)
```

```python
# Apply standard filters
# keep rows where impressions_90d > 0 and content_age_days >= 90
filtered_df = df[(df['impressions_90d'] > 0) & (df['content_age_days'] >= 90)].copy()

# Dedupe by content_id (keeping the first occurrence)
filtered_df.drop_duplicates(subset=['content_id'], inplace=True)
```

```python
# Calculate metrics
total_pages_after_filter = len(filtered_df)
declining_pages = filtered_df[filtered_df['trend_direction'] == 'down']
num_declining_pages = len(declining_pages)
percent_declining_pages = (num_declining_pages / total_pages_after_filter) * 100 if total_pages_after_filter > 0 else 0

total_impressions_all = filtered_df['impressions_90d'].sum()
total_impressions_declining = declining_pages['impressions_90d'].sum()
share_impressions_declining = (total_impressions_declining / total_impressions_all) * 100 if total_impressions_all > 0 else 0

avg_word_count_declining = declining_pages['word_count'].mean()
avg_word_count_non_declining = filtered_df[filtered_df['trend_direction'] != 'down']['word_count'].mean()

avg_position_declining = declining_pages['avg_position'].mean()
avg_position_non_declining = filtered_df[filtered_df['trend_direction'] != 'down']['avg_position'].mean()
```

```python
print(f"After filtering and deduping, there are {total_pages_after_filter} unique content pages.")
print(f"{num_declining_pages} pages ({percent_declining_pages:.2f}%) are currently labeled 'down'.")
print(f"These declining pages represent {share_impressions_declining:.2f}% of total impressions.")
print(f"Average word count for declining pages: {avg_word_count_declining:.2f}")
print(f"Average word count for non-declining pages: {avg_word_count_non_declining:.2f}")
print(f"Average position for declining pages: {avg_position_declining:.2f}")
print(f"Average position for non-declining pages: {avg_position_non_declining:.2f}")
```
